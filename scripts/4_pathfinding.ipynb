## Overview of cancer children & if there is model

```python
with open("../data/cancer_children.json") as f:
    children = json.load(f)
```

```python
terms = children['_embedded']['terms']
ids = [t['short_form'] for t in terms]
names = [t['label'] for t in terms]
models = []
for id in ids:
    match = stats['ID'].str.replace(":", "_").str.contains(id).any()
    models.append(match)
```

```python
cancer_children_1 = {'ids': ids, 'names': names, 'models': models}
cancer_children_1 = pd.DataFrame(cancer_children_1)
cancer_children_1
```

## Running mouse data on cancer & Pick out the children cancer with largest number of predictions ($log(prob) > 1$)

```python
# # Preprocess
# subprocess.run([
#     "python", "../src/preprocess.py",
#     "-input", "../demo/metadata_mouse_title.tsv",
#     "-out", "../demo/metadata_mouse_title_processed.tsv"
# ])

# # Generate embeddings
# subprocess.run([
#     "python", "../src/embedding_lookup_table.py",
#     "-input", "../demo/metadata_mouse_title_processed.tsv",
#     "-out", "../demo/metadata_mouse_title_embedding.npz"
# ])

# Predict
for index, row in cancer_children_1.iterrows():
    if row['models']:
        subprocess.run([
            "python", "../src/predict.py",
            "-input", "../demo/metadata_mouse_title_processed.tsv",
            "-id", "../demo/metadata_mouse_ID.tsv",
            "-input_embed", "../demo/metadata_mouse_title_embedding.npz",
            "-train_embed", "../data/disease_desc_embedding.npz",
            "-model", f"../bins/{row['ids']}__model.pkl",
            "-out", "../results/cancer_children_mouse/"
        ])
```

```python
files = [f for f in os.listdir("../results/cancer_children_mouse") if f.endswith(".csv")]

dfs = []
for f in files:
    df = pd.read_csv(os.path.join("../results/cancer_children_mouse", f))
    df['source'] = f.replace("__preds.csv", "")
    dfs.append(df)

all_preds = pd.concat(dfs, ignore_index=True)
all_preds.to_csv(
    os.path.join("../results/cancer_children_mouse", "all_cancer_predictions.csv"),
    index=False
)
print("Saved all_cancer_predictions.csv")
```

```python
all_preds[all_preds['log2(prob/prior)']>0].groupby('source').size().reset_index(name='count').sort_values('count', ascending=False)
```

## Choose 0000621 - immune system cancer & check children nodes distribution

```python
# ! wget https://www.ebi.ac.uk/ols4/api/ontologies/mondo/terms/http%253A%252F%252Fpurl.obolibrary.org%252Fobo%252FMONDO_0000621/children
```

```python
with open("../data/0000621_children.json") as f:
    children = json.load(f)

terms = children['_embedded']['terms']
ids = [t['short_form'] for t in terms]
names = [t['label'] for t in terms]
models = []
for id in ids:
    match = stats['ID'].str.replace(":", "_").str.contains(id).any()
    models.append(match)

cancer_children_2 = pd.DataFrame({'ids': ids, 'names': names, 'models': models})
cancer_children_2
```

```python
# Predict
for index, row in cancer_children_2.iterrows():
    if row['models']:
        subprocess.run([
            "python", "../src/predict.py",
            "-input", "../demo/metadata_mouse_title_processed.tsv",
            "-id", "../demo/metadata_mouse_ID.tsv",
            "-input_embed", "../demo/metadata_mouse_title_embedding.npz",
            "-train_embed", "../data/disease_desc_embedding.npz",
            "-model", f"../bins/{row['ids']}__model.pkl",
            "-out", "../results/0000621_children_mouse/"
        ])

files = [f for f in os.listdir("../results/0000621_children_mouse") if f.endswith(".csv")]

dfs = []
for f in files:
    df = pd.read_csv(os.path.join("../results/0000621_children_mouse", f))
    df['source'] = f.replace("__preds.csv", "")
    dfs.append(df)

all_preds = pd.concat(dfs, ignore_index=True)
all_preds.to_csv(
    os.path.join("../results/0000621_children_mouse", "all_cancer_predictions.csv"),
    index=False
)
print("Saved all_cancer_predictions.csv")
```

```python
all_preds[all_preds['log2(prob/prior)']>0].groupby('source').size().reset_index(name='count').sort_values('count', ascending=False)
```
