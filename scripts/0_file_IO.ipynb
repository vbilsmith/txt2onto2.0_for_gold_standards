## metadata

```python
# Import
GREIN_data = pd.read_csv("../demo/GREIN_data.csv")
```

```python
# Import
with open("../demo/metadata_human.json") as f:
    metadata_human = json.load(f)
with open("../demo/metadata_mouse.json") as f:
    metadata_mouse = json.load(f)

# Export documents from metadata
ids = []
species = []
titles = []
summaries = []

for gse in metadata_human:
    ids.append(gse)
    species.append(metadata_human[gse]["Organism"])
    titles.append(metadata_human[gse]["Title"])
    summaries.append(metadata_human[gse]["Summary"])

for gse in metadata_mouse:
    ids.append(gse)
    species.append(metadata_mouse[gse]["Organism"])
    titles.append(metadata_mouse[gse]["Title"])
    summaries.append(metadata_mouse[gse]["Summary"])

data = {'ids':ids, 'species':species, 'titles':titles, 'summaries':summaries}
metadata = pd.DataFrame(data).sort_values('ids')
metadata['documents'] = metadata['titles'] + metadata['summaries']

metadata.to_csv("../demo/metadata_concat.tsv", sep="\t", index=False)
pd.DataFrame(metadata['ids']).to_csv("../demo/metadata_concat_ID.tsv", sep="\t", index=False, header=False)
pd.DataFrame(metadata['documents']).to_csv("../demo/metadata_concat_document.tsv", sep="\t", index=False, header=False)
```
