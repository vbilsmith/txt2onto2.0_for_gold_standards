```python
import pandas as pd
import json
import subprocess
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
stats = pd.read_csv("../data/disease_model_stats.csv")
model_ids = stats['ID'].str.replace(":", "_")
```

```python
mouse_title_53599 = "Analysis of gene expression changes in RN2 acute myeloid leukemia (AML) cells at 24hrs and 48hrs after Ddx5 knockdown."

subprocess.run([
    "python", "../src/preprocess.py",
    "-input", "../demo/metadata_mouse_title_53599.tsv",
    "-out", "../demo/metadata_mouse_title_processed_53599.tsv"
])

subprocess.run([
    "python", "../src/embedding_lookup_table.py",
    "-input", "../demo/metadata_mouse_title_processed_53599.tsv",
    "-out", "../demo/metadata_mouse_title_embedding_53599.npz"
])

for id in model_ids:
    subprocess.run([
        "python", "../src/predict.py",
        "-input", "../demo/metadata_mouse_title_processed_53599.tsv",
        "-id", "../demo/metadata_mouse_ID_53599.tsv",
        "-input_embed", "../demo/metadata_mouse_title_embedding_53599.npz",
        "-train_embed", "../data/disease_desc_embedding.npz",
        "-model", f"../bins/{id}__model.pkl",
        "-out", "../results/GSE53599"
    ])
```
