```python
import json
import csv

def json_to_tsv(input_json,
                out_text1="title_summary.tsv",
                out_text2="title_summary_design.tsv",
                out_ids="gse_ids.tsv"):
    
    with open(input_json, "r", encoding="utf-8") as f:
        data = json.load(f)

    text1_rows = []
    text2_rows = []
    id_rows = []

    # data 是 dict：key = GSE id
    for gse_id, record in data.items():
        title = record.get("Title", "").strip()
        summary = record.get("Summary", "").strip()
        design = record.get("Overall design", "").strip()

        text1_rows.append([f"{title} {summary}"])
        text2_rows.append([f"{title} {summary} {design}"])
        id_rows.append([gse_id])

    with open(out_text1, "w", newline="", encoding="utf-8") as f:
        csv.writer(f, delimiter="\t").writerows(text1_rows)

    with open(out_text2, "w", newline="", encoding="utf-8") as f:
        csv.writer(f, delimiter="\t").writerows(text2_rows)

    with open(out_ids, "w", newline="", encoding="utf-8") as f:
        csv.writer(f, delimiter="\t").writerows(id_rows)

    print("Done.")
```

```python
json_to_tsv("metadata_human.json",
            "human_title_summary.tsv",
            "human_title_summary_design.tsv",
            "human_ID.tsv")
```

```python
json_to_tsv("metadata_mouse.json",
            "mouse_title_summary.tsv",
            "mouse_title_summary_design.tsv",
            "mouse_ID.tsv")
```
